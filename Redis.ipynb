# 1&rpar; What are redis and MongoDB?










##1.1 What type of database is each?

REmote DIctionary Server (redis) is an open source , in-memory, NoSQL key-value store that is primarly used as an application cache or quick-response database.
It stores data as simple pair of keys and values, similar to a python dictionary.

MongoDB is an open source, nonrelational, document-oriented database management system that uses flexible documents instead of tables and rows to process and store various forms of data. It stores values (documents) in a BSON (binary JSON) format for efficiency.

##1.2 How does each leverage types of storage/memory?

MongoDB primary stores its data on a persistent Hard Drive (Disk). It also pulls frequently accessed data into RAM to speed up reads. It can store massive amounts of data for cheap. It is a non-volatile datastore.

Redis store all of its data directly in RAM (in memory). Becuase reading from RAM is exponentially faster than reading from a disk. But also the reason that its size is limited. Redis is a volatile datastore.



##1.3 How They Are the Same vs. Different?

SIMILARITIES :

*  Both are NoSQL databases. Both are designed to handle unstructured or semi-structured data and can scale easily.


DIFFERENCES:

*   Volatility/Persistence: Redis data lives in RAM; if the system crashes, data is lost unless explicitly configured with persistence (Snapshotting/AOF). MongoDB writes data to disk, ensuring permanence.

*   Performance: Redis offers higher throughput and lower latency than MongoDB due to in-memory storage.

*   Use Cases: Redis is ideal for caching, session management, and real-time analytics. MongoDB is suited for storing large volumes of data, content management, and complex queries.

*   Data Models: Redis represents data simply as keys mapped to values, which can be strings or containers of strings like hashes and lists. In contrast, MongoDB values are documents that can be nested to an arbitrary depth.


While Redis can be configured to persist data to disk, it is not inherently designed as a persistent, durable database like MongoDB.


##1.4 When would each typically be used?

*  MongoDB &rarr;  It is suitable for applications with semi-structured data and quickly evolving systems. In data science, it is also used as a NoSQL Model Registry for deploying machine learning models. It is also used when you need to store complex, nested data permanently.

Example: Perfect for primary databases for web/mobile apps, content  management systems, e-commerce catalogs, and storing user profiles.

*   Redis &rarr; It is commonly used as a caching layer sitting between the user/application and the main database to provide much faster lookups. It is a popular choice for distributed caches.

Exmaple: session management (keeping users logged in), leaderboards, and real-time analytics.





##1.5 What are the tradeoffs for each?

MongoDB: It is not well suited for heavy, complex transactional systems, such as banking applications. Additionally, filtering, aggregating, and ad hoc querying are generally more difficult than in relational databases.

Redis:  It has no filtering capabilities, meaning you cannot query just a part of a value; you must fetch the entire thing. Also, if you do not know the exact key you need, you are forced to iterate over the keys, which eliminates its performance advantage.

## 3.1 How is this setup same or different from SQL?

The similarities:

*   Data is still organized into distinct, logical groupings. Instead of relational "tables" and "Rows", we can MongoDB "Collections" and "Documents".

*  We are interacting with the database programatically using pymongo similar to sqlite3, connecting to a server, and executing commands to read and write data.

The differences:

*   No predefines SQL schema: in SQL, I had to crite DDL queries to strictly create, define column names and datatype before inserting anything in it. In MongoDB we can sompley insert the python dictionaries directly, and Mongo accepts the structure dynamically.
*   No JOINs: In SQLite, I created an OrderItems tables to handle the many-to-many relationship between inventory and order tables. In MongoDB, I could denormalize by embedding the line items directly into orders documents as an array. This eliminates the need for JOIN queries, as retreiving one order document instantly retrieves all its line items.

#  5&rpar; Caching with Redis

Why a cache could be useful and how redis would work with Mongo:

Cache is a temporary storage layer that holds frequently accessed data so applications won't have to fetch it from the slower, primary database everytime. Redis is an in-memory key-value store and thus it is exceptionally fast, and executes in O(N) time. By placing Redis as a cache layer between the application and MongoDB, we can intercept common queries like ("What is the total number of movies?"). If the answer is in Redis, we can immediately return it or else, we can query it from MongoDB and store the result for the next person and return it.

## 5.1 Tradeoffs

*  Memory limitations : RAM is much more expensive and limited in size compared to disk storage, so you cannot cache your entire MongoDB database.
*  No Filtering / Complex queries : Redis stores data as simple key-value pairs. You cannot query a part of a value that you need. That is, even if you require a subset of the data of the documnent, you will have to fetch the whole value and filter it in the application code.
*  Adding a cache layer introduces another system to manage, monitor and keep synchronized.





## 5.2 How to model / setup data in cache?

*   I should name the keys carefully in a way that iterating through the cache shouldn't eliminate its performance advantage.
*   The cache requires rules on when to delete old data and fetch fresh data.





## 5.3 A simple, locally running Redis cache.

```python
import redis
import json
redis_client = redis.Redis(host = 'localhost', port=6379)
redis_client.mset({'total_movies_count':total_movies_count,
'oldest_movie':json.dumps(oldest_movie),
'bad_but_awarded':json.dumps(bad_but_awarded)})

values = redis_client.mget(['total_movies_count','oldest_movie','bad_but_awarded'])
decoded_values = [val.decode('utf-8') for val in values]
val_total = decoded_values[0]
val_oldest = json.loads(decoded_values[1])
val_bad = json.loads(decoded_values[2])

print(f"count of total movies is {val_total}")
print(f"oldest movies is {val_oldest}")
print(f"movie won awards but with less than 1.7 IMDB rating {val_bad}")
```

## 5.4 Concerns as a Data Scientist using cached data

As a data scientist, doing analysis on top of cached data pose the following risks :-



*   Data becomes stale: The cache might not reflect the most recent updates made to the primary MongDB data base. Analysing stale data can lead to inaccuracies and inconsistencies if chache eviction policies are not set up properly.
*  A cache is strictly a temporary layer and not a full database, it hold a temporary subset of data that is most frequently accessed. Performing analysis on this data means inherently being biased to the popular data. The model would be completly ignoring the rare or older events, leading to skewed preditions.
* Redis is a key-value datastore with a simple data model. We cannot filter out or query any specific data we want, we will always have to fetch the entire object. And if we don't know the exact keys to fetched, we will be forced to iterate through all the entires, which completly eliminates the performance advantage opf caching in the first place.

## Are there any outliers?

Let us try to find the outliers in population count of countries to find the most populous and least populous countries in the world.


The global median population is in the millions. To find the statistical outliers, I will write a MongoDB query to find the countries in the right tail (population > 1,000,000,000) and the extreme left tail (population < 1,000).

```python
print("\n--- 🌍 Outlier Countries (Population Extremes) ---")
outliers = list(
    countries_db.countries.find(
        {"$or": [
            {"population": {"$gt": 1000000000}},
            {"population": {"$lt": 1000}}
        ]},
        {"name.common": 1, "population": 1, "region": 1, "_id": 0}
    ).sort("population", -1)
)

for country in outliers:
    name = country.get('name',{}).get('common')
    population = country.get('population',0)
    print(f"Country: {name} | population: {population:,}")
```

To fully implement cache for our countries datastore, below is the workflow.

1. The application asks Redis for the data first
2. If Redis has the required data in the cache, it returns it instantly - CACHE HIT
3. If Redis does not have it (CACHE MISS), the application queries MongoDB, get the answer, saved it to Redis for the next time, and then returns the data.
